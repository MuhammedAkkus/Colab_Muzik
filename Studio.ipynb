```python
import subprocess
from pathlib import Path
repo = Path('/content/Colab_Muzik_Studio')
if not repo.exists():
    subprocess.run(['git', 'clone', '--depth', '1', 'https://github.com/MuhammedAkkus/Colab_Muzik.git', str(repo)], check=True)
else:
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
subprocess.run(['python', '-m', 'pip', 'install', '-q', '-r', str(repo / 'requirements.txt')], check=True)
subprocess.run(['apt-get', 'install', '-y', '-qq', 'ffmpeg'], check=True)
```

```python
import os
from datetime import datetime
from zoneinfo import ZoneInfo
today = datetime.now(ZoneInfo('Europe/Istanbul')).date().isoformat()
catalog = repo / 'catalog'
if (catalog / today / 'manifest.json').exists():
    print('Existing GitHub daily attempt preserved. No duplicate API calls.')
else:
    from google.colab import userdata
    free_project_confirmed = True # Set False if billing is ever enabled
    assert free_project_confirmed, 'Billing-disabled free project required'
    catalog = Path('/content/studio_catalog')
    environment = dict(os.environ, GEMINI_API_KEY=userdata.get('GEMINI_API_KEY'))
    try:
        result = subprocess.run(['python', 'studio.py', 'run', '--catalog', str(catalog)], cwd=repo, env=environment)
        print('Status code:', result.returncode, '(0 completed/previous attempt; 2 rejected; 1 failed)')
    finally:
        environment.pop('GEMINI_API_KEY', None)
```

```python
import json
import shutil
from IPython.display import Audio, display, Markdown
from google.colab import files
for folder in sorted(catalog.glob('*')):
    manifest = folder / 'manifest.json'
    if manifest.exists():
        print(folder.name, json.loads(manifest.read_text())['status'])
    lyrics = folder / 'lyrics.md'
    if lyrics.exists():
        display(Markdown(lyrics.read_text()))
    master = folder / 'instrumental.mp3'
    if not master.exists():
        master = folder / 'preview.mp3'
    if master.exists():
        display(Audio(filename=str(master)))
if catalog.exists():
    archive = shutil.make_archive('/content/Colab_Muzik_Studio', 'zip', catalog)
    files.download(archive)
```
